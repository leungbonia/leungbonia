#Python Basic Definitions

Objects: A data record with fields; an instance of a class: `type()`

Functions: a reusable piece of code that performs a specific task

Variables: A defined name that references an object, e.g. `id()`. Can assign function to variable

Class: A template for creating objects (records). Data type. Can build your own class.

Methods:A function of an object. A function defined inside a class that operates on its objects.

Attributes:A variable of an object. A field in a record, defined by its class. No `()`
